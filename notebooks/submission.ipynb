```python
import os

def list_subfolders(folder_path):
    """Returns a list of all sub-folder names in the given folder."""
    return [name for name in os.listdir(folder_path) if os.path.isdir(os.path.join(folder_path, name))]

# Example usage
folder_path = DATA_DIR / "raw" / "test"
subfolders = list_subfolders(folder_path)
```

```python
import h5py

def get_scenario_ids(h5_filepath):
    """Extracts all scenario_ids from the given .h5 file."""
    scenario_ids = []
    
    with h5py.File(h5_filepath, "r") as hf:
        for idx_key in hf.keys():  # Iterate through all keys in the file
            scenario_ids.append(hf[idx_key].attrs["scenario_id"])
    
    return scenario_ids

# Example usage
h5_filepath = DATA_DIR / "test.h5"
scenario_ids = get_scenario_ids(h5_filepath)
```
